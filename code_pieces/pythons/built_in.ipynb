```python
# 表示不同的 状态

from enum import Enum, auto 

# 
class State(Enum):
    PENDING = auto()
    RUNNING = auto()
    SUCCESS = auto() 

def handle_state(state: State):
    if state == State.PENDING:
        print("pending")
    elif state == State.RUNNING:
        print("running")
    elif state == State.SUCCESS:
        print("success")
# 不该用的场景：枚举值需要运行时动态注册（比如插件系统从配置文件加载状态），enum 的静态特性反而是限制。
```
