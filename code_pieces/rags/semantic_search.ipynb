# 语义检索：embedding + cosine similarity

语义检索适合自然语言问题、同义表达和用户不知道准确术语的场景。它不替代 BM25；生产 RAG 里通常用 hybrid retrieval。

```python
from pathlib import Path
import sys

sys.path.append(str(Path.cwd()))
sys.path.append(str(Path.cwd() / "snippets" / "rags"))
sys.path.append(str(Path.cwd().resolve().parent))

docs = [
    "差旅政策规定了机票、酒店和报销流程。",
    "办公家具采购包括桌椅、显示器支架和会议室设备。",
    "RAG 系统通常包括文档切分、向量召回、重排序和答案生成。",
    "BM25 对错误码、接口名、专有名词这类精确关键词很有效。",
]
query = "员工出差费用怎么报销"
```

```python
# Install first if needed:
# pip install sentence-transformers scikit-learn numpy
from semantic_search import SemanticRetriever, load_sentence_transformer

model = load_sentence_transformer("BAAI/bge-small-zh-v1.5")
doc_embeddings = model.encode(docs, normalize_embeddings=True)
query_embedding = model.encode([query], normalize_embeddings=True)

retriever = SemanticRetriever(docs, doc_embeddings)
retriever.search(query_embedding, top_k=3)
```

## 经验

- 如果专有名词、错误码召回差，把 BM25 合进来。
- 如果 top-k 有答案但排序不稳，加 reranker。
- 如果答案引用上下文不完整，先调 chunk，而不是先换大模型。
