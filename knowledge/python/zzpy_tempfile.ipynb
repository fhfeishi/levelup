# subprocess

subprocess是Python用于生成新进程、链接其 输入/输出/错误管道 并获取其返回码的标准模块。它是用来替换旧的 os.system 和 os.spawn* 的强大工具

1. 核心方法： run  Popen

-  subprocess.run() (推荐用法)
    -  特点：阻塞式。它会运行命令并等待命令完成，然后返回一个 CompletedProcess 对象。
    - 适用场景：你只需要运行一个命令，并不关心它运行时的实时输出，只在乎最后的结果（成功/失败、最终输出）。

- subprocess.Popen() (高级用法)
    - 特点：非阻塞式（初始化时）。它会启动进程，然后你可以继续执行 Python 代码，或者通过 communicate() 方法与进程交互。它提供了更细粒度的控制。
    - 适用场景：需要实时获取输出流、需要并发执行、或者通过 stdin 持续给子进程喂数据时。
