````python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr 
    pivot = arr[len(arr)//2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)
    
# 测试一下
test_arr = [3, 6, 8, 10, 1, 2, 1]
print(quick_sort(test_arr))
# 输出: [1, 1, 2, 3, 6, 8, 10]
````

````python
def quick_sort(arr):
    # 基准情况：如果数组为空或只有一个元素，直接返回
    if len(arr) <= 1:
        return arr 
    
    # 选择基准值 (Pivot)
    pivot = arr[len(arr) // 2]
    
    # ---------------- 补全部分开始 ----------------
    
    # 1. 分区：将数组分为三部分
    # 比基准值小的元素
    left = [x for x in arr if x < pivot]
    
    # 等于基准值的元素 (处理重复值)
    middle = [x for x in arr if x == pivot]
    
    # 比基准值大的元素
    right = [x for x in arr if x > pivot]
    
    # 2. 递归：对左右两边继续排序，并连接结果
    return quick_sort(left) + middle + quick_sort(right)
    
    # ---------------- 补全部分结束 ----------------

# 测试一下
test_arr = [3, 6, 8, 10, 1, 2, 1]
print(quick_sort(test_arr))
# 输出: [1, 1, 2, 3, 6, 8, 10]
````

````python
def dfs(graph, start):
    # 1. 初始化：使用集合记录已访问节点，使用列表作为栈
    visited, stack = set(), [start]
    
    # 2. 只要栈不为空，就继续循环
    while stack:
        # 弹出栈顶元素（后进先出 LIFO）
        vertex = stack.pop()
        
        # 3. 如果该节点未被访问过
        if vertex not in visited:
            # 标记为已访问
            visited.add(vertex)
            print(vertex, end=" ")  # 处理节点（例如打印）
            
            # 4. 将邻居节点压入栈中
            # 注意：这里使用 reversed 是为了保证访问顺序是从左到右
            # 如果不介意访问顺序，可以直接 stack.extend(graph[vertex])
            stack.extend(reversed(graph[vertex]))
            
    return visited
````
