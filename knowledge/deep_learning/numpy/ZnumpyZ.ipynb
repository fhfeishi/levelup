## np.clip()

- np.clip() 是 NumPy 提供的一个用于 限制数值范围 的函数，常用于深度学习和数值计算中，以 防止溢出或确保数据在合理范围内。
- np.clip() 语法
    ```python
    np.clip(a, a_min, a_max, out=None)
    ## 参数 
    # a：输入的 NumPy 数组。
    # a_min：最小值（可以是标量或数组-match）。
    # a_max：最大值（可以是标量或数组-match）。
    # out（可选）：如果提供，则存储结果，而不创建新的数组。
    ## 返回值
    # 返回一个数组，其中 a 的值被限制在 [a_min, a_max] 之间：
    #   如果 a 的值小于 a_min，则替换为 a_min。
    #   如果 a 的值大于 a_max，则替换为 a_max。
    #   如果 a_min=None，则只应用 a_max 的限制。
    #   如果 a_max=None，则只应用 a_min 的限制。
    ```

````python
arr = np.array([1,2,3,4,5])
arr = np.clip(arr, 1,3)
arr
````

- np.clip() 在深度学习中的应用
    - 防止 log(0) 错误：在交叉熵损失计算时，我们需要 log(y_pred)，但如果 y_pred=0，会导致 log(0) = -∞，产生 NaN 错误。因此，我们使用 np.clip() 限制 y_pred
    - 防止梯度爆炸：在梯度更新时，梯度可能会变得非常大，导致参数更新过大，模型训练不稳定。我们可以用 np.clip() 限制梯度值

````python
y_pred = np.array([0.0, 0.5, 1.0])
y_pred = np.clip(y_pred, 1e-12, 1.0)  # 限制在 [1e-12, 1.0]
print(y_pred)  # [1.e-12 5.e-01 1.e+00]

gradients = np.array([-100, -5, 0, 5, 100])
clipped_gradients = np.clip(gradients, -10, 10)
print(clipped_gradients)  # [-10  -5   0   5  10]
````

* **np.clip() vs np.maximum() 和 np.minimum()** <br>
np.clip(arr, a_min, a_max) =等价= np.minimum(np.maximum(arr, a_min), a_max)

````python
arr = np.array([-5, 0, 5, 10, 15])
print(np.clip(arr, 0, 10))   # [ 0  0  5 10 10 ]
print(np.minimum(np.maximum(arr, 0), 10))  # [ 0  0  5 10 10 ]
````

- 有符号整数 (Signed Integer)
    - 定义: 有符号整数可以表示正数、负数和零。
    - 表示方法: 通常使用最高位（最左边的位）作为符号位，0 表示正数，1 表示负数。其余位表示数值的大小。
    - 范围: 例如，对于 8 位有符号整数 (np.int8)，其范围是 -128 到 127。
    - 用途: 适用于需要表示负数的场景，例如温度变化、财务数据等。
- 无符号整数 (Unsigned Integer)
    - 定义: 无符号整数只能表示非负数（包括零）。
    - 表示方法: 所有位都用于表示数值的大小，没有专门的符号位。
    - 范围: 例如，对于 8 位无符号整数 (np.uint8)，其范围是 0 到 255。
    - 用途: 适用于不需要表示负数的场景，例如图像处理中的像素值、计数器等。

- 转换注意事项
    - 从有符号到无符号: 如果将有符号整数转换为无符号整数，负数会被解释为非常大的正数。例如，np.int8 的 -1 转换为 np.uint8 后会变成 255。
    - 从无符号到有符号: 如果将无符号整数转换为有符号整数，大于等于最大正数的值会被解释为负数。例如，np.uint8 的 255 转换为 np.int8 后会变成 -1。

数据类型解释

- np.uint8:
    - 描述: 无符号 8 位整数。

    - 描述: 无符号 16 位整数。
    - 范围: 0 到 65535。
    - 用途: 用于需要更大范围的无符号整数的场景，例如某些科学数据或高精度图像处理。
- np.int8:
    - 描述: 有符号 8 位整数。
    - 范围: -128 到 127。
    - 用途: 用于需要小范围有符号整数的场景。
- np.int16:
    - 描述: 有符号 16 位整数。
    - 范围: -32768 到 32767。
    - 用途: 用于需要更大范围的有符号整数的场景。
- np.int32:
    - 描述: 有符号 32 位整数。
    - 范围: -2147483648 到 2147483647。
    - 用途: 用于大多数通用计算场景。
- np.int64:
    - 描述: 有符号 64 位整数。
    - 范围: -9223372036854775808 到 9223372036854775807。
    - 用途: 用于需要非常大范围的有符号整数的场景，例如某些科学计算或大数据处理。
- np.float32:
    - 描述: 单精度浮点数。
    - 范围: 大约 ±3.4e38，精度约为 7 位小数。
    - 用途: 用于需要浮点数但内存占用较小的场景。
- np.float64:
    - 描述: 双精度浮点数。
    - 范围: 大约 ±1.8e308，精度约为 16 位小数。
    - 用途: 用于大多数需要高精度浮点数的场景。

- Python的列表和NumPY数组有什么区别？
    - NumPy 提供了大量快速有效的方法来创建数组并在其中操作数值数据。虽然 Python 列表可以在单个列表中包含不同的数据类型，但 NumPy 数组中的所有元素都应该是同质的。如果数组不是同质的，那么在数组上执行的数学运算将极其低效。
    - NumPy 数组比 Python 列表更快、更紧凑。数组占用的内存更少，使用起来更方便。NumPy 使用更少的内存来存储数据，并且它提供了指定数据类型的机制。这允许进一步优化代码。

与其他 Python 容器对象一样，可以通过索引或切片来访问和修改数组的内容。与典型的容器对象不同，不同的数组可以共享相同的数据，因此对一个数组所做的更改可能会在另一个数组中可见。

- NumPY 的副本和视图
    - 副本是一个数据的完整的拷贝，如果我们对副本进行修改，它不会影响到原始数据，物理内存不在同一位置。调用ndarray类的copy（）函数会产生一个副本，类似的还有python中的deepcopy（）函数。
    - 视图是数据的一个别称或引用，通过该别称或引用亦便可访问、操作原有数据，但原有数据不会产生拷贝。调用ndarray类的view（）函数会创建数组的视图，切片操作也会返回原数据的视图。

````python
print(f"{b = }")  # b = array([[1, 2],[2, 3],[4, 5]])

bb = b # 无复制
b_view = b.view()  # 视图或浅拷贝
b_copy = b.copy()  # 副本或深拷贝
b_slice = b[0:1]   # 切片， b的(部分)视图 

print(f"{id(b) = }")
print(f"{id(bb) = }")
print(f"{id(b_view) = }")
print(f"{id(b_copy) = }")
print(f"{id(b_slice) = }")

print(f"{ (id(b) == id(bb)) = }")     # True
print(f"{(id(b) == id(b_view)) = }") # False
print(f"{  (id(b) == id(b_copy)) = }")  # False
print(f"{(id(b) == id(b_slice)) = }")  # False
````

````python
# b 改变： b[0] + 1  # 这里其实就应用了numpy的广播机制，将1调整为b[0]的形状[1,1]，然后加
b[0]+=1
print(f"{b = }")  # change

print(f"{bb =}")  # change 
print(f"{b_view = }")  # change b的视图 跟python的copy(浅拷贝)一样，嵌套结构内的元素会跟着改变 
print(f"{b_slice = }") # change b的一部分视图

print(f"{b_copy = }")  # --
````

````python
bb[0] -= 1
print(f"{bb =}")  # change

print(f"{b = }")  # change
print(f"{b_view = }")  # change 
print(f"{b_slice = }") # change

print(f"{b_copy =}")  # --
````

````python
b_view[0] += 2  # 视图
print(f"{b_view = }") # change

print(f"{bb =}")  # change
print(f"{b = }")  # change
print(f"{b_slice = }") # change

print(f"{b_copy =}")  # --
````

````python
b_slice -= 8  # 视图
print(f"{b_slice = }") # change

print(f"{bb =}")  # change
print(f"{b = }")  # change
print(f"{b_view = }")  # change
print(f"{b_copy =}")  # --
````

````python
b_copy[0] += 2
print(f"{b_copy =}")  # change

print(f"{bb =}")  # --
print(f"{b = }")  # --
print(f"{b_view = }")  # --
print(f"{b_slice = }") # --
````

- 赋值（=）不会创建副本，而是让变量指向相同的数据。修改其中一个变量的值会影响另一个，因为它们共享同一块内存。。 
- ndarray.view() 创建一个新的数组对象，但数据仍然共享相同的物理内存。view() 不能修改原数组的形状，但 reshape() 可以重新组织视图的数据。
- 切片（slice）通常返回视图，而不是副本。修改切片的数据会影响原数组

## 广播机制
- 当数组形状不同但可以匹配时，NumPy 自动扩展较小的数组
    - 从右向左对齐形状（即最后一个维度先对齐）
    - 如果某个维度的大小相同，或其中一个为 1，则可以广播。
    - 如果维度不匹配且都不为 1，则无法广播，会报错


```python
# arr 可以匹配那些shape的数组  -- 支持广播
ndarray.shape = (a, b, c) 
    - np.array([int])  # shape (1,)
    - np.array([...])  # shape (c,)
    - shape (1,c)
    - shape (b,c)
    - shape(1,b,c)
    - shape(a,b,c)
    - shape(1,a,b,c)
    - shape(...,a,b,c)


````python
1+ np.array([[0,0],[0,0]])
````

````python
arrx = np.zeros((2,3,4))  # shape(a,b,c)

arrx   + np.ones_like([2])     # (1,)
arrx   + np.arange(4)          # (c,)
arrx   + np.zeros((1,4))       # (1,c)
arrx   + np.zeros((3,4))       # (b,c)
arrx   + np.zeros((1,3,4))     # (1,b,c)
arrx   + np.zeros((2,3,4))     # (a,b,c)
arrx   + np.zeros((1,2,3,4))   # (1,a,b,c)
arrx   + np.zeros((1,2,2,3,4)) # (..., a, b, c)
````
