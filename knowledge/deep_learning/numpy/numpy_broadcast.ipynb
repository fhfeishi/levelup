```python
import numpy as np

arr1 = np.array([[1,1], [2,2]]) # 向量，(2,2)矩阵
arr2 = np.array([2,2])  # 标量，一列数   
arr3 = np.array([1,2])  # 标量，一列数   
""" 
arr1  1, 1    arr2  2    arr3  1
      2, 2          2          2
"""
print(f"{np.matmul(arr1,arr2) = }") # array([4, 8])  
# arr2  --广播->  (2,1)列向量 
print(f"{(np.matmul(arr1,arr2)).shape = }") # (2,)  
#  广播: np.matmul(矩阵,标量) -> 标量 形状跟输入的标量一样
print(f"{np.matmul(arr1,arr3) = }") # array([3, 6])  
# arr3  --广播->  (2,1)列向量 
```
