```python
import mindspore
from mindspore import ops

# 确保在PyNative模式
mindspore.set_context(mode=mindspore.PYNATIVE_MODE)

# 创建张量并显式转换类型
a = ops.ones((1, 2))
a = mindspore.Tensor(a)  # 转换为常规Tensor

b = ops.zeros((1, 2))
b = mindspore.Tensor(b)  # 转换为常规Tensor

c = ops.zeros_like(b, dtype=b.dtype)  # 显式指定dtype
d = ops.ones_like(a)
e = ops.one_hot(indices=a.astype(mindspore.int32), depth=4, on_value=0., off_value=3.)  # 确保indices为整型
f = ops.full(a.shape, 3)
g = ops.full_like(c, 4)
h = ops.eye(4)
```

```python
a = mindspore.ops.ones((1,2))
# print(f"{type(a) = }")  # <class 'mindspore.common._stub_tensor.StubTensor'>
b = mindspore.ops.zeros((1,2))
c = mindspore.ops.zeros_like(b, dtype=b.dtype)    
d = mindspore.ops.ones_like(a, dtype=a.dtype)
e = mindspore.ops.one_hot(indices=a, depth=4, on_value=0., off_value=3.)
f = mindspore.ops.full(a.shape, 3)
g = mindspore.ops.full_like(c, 4, dtype=c.dtype)
h = mindspore.ops.eye(4)
```
