# 关键词检索：倒排索引 / TF-IDF / BM25

关键词检索适合错误码、函数名、字段名、政策条款、产品型号等需要精确匹配的场景。真实 RAG 中，它通常和语义检索一起组成 hybrid retrieval。

```python
from pathlib import Path
import sys

sys.path.append(str(Path.cwd()))
sys.path.append(str(Path.cwd() / "snippets" / "rags"))
sys.path.append(str(Path.cwd().resolve().parent))

docs = [
    "RAG 检索包括关键词搜索、语义搜索、混合检索。",
    "BM25 适合精确词匹配、术语、错误码、函数名。",
    "语义搜索适合同义表达，例如报销制度和差旅政策。",
    "reranker 通常放在召回后，对候选 chunk 重新排序。",
]
query = "BM25 错误码 检索"
```

```python
from keyword_search import build_inverted_index, search

index, counts = build_inverted_index(docs)
search(query, docs, index, counts, top_k=3)
```

```python
# BM25 usually gives a stronger lexical baseline.
# Install first if needed: pip install rank-bm25
from bm25 import build_bm25, search as bm25_search

bm25 = build_bm25(docs)
bm25_search(query, docs, bm25, top_k=3)
```
