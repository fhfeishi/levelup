```python

class zzcls:
    name: str = "zzcls"   # cls attribute
    cls_name: str = "zzcls_cls" # cls attribute
    
    def __init__(self, name: str):
        self.name = name   # instance attribute  

# instance attribute may cover cls attribute 
s1 = zzcls("s1")
s2 = zzcls("s2")

print(s1.name) # s1 
print(s2.name) # s2

print(s1.cls_name) # zzcls_cls
print(s2.cls_name) # zzcls_cls

# change zzcls cls_name
print(zzcls.cls_name) # zzcls_cls
zzcls.cls_name = "zzcls_cls_new"

print(zzcls.cls_name) # zzcls_cls_new
print(s1.cls_name) # zzcls_cls_new
print(s2.cls_name) # zzcls_cls_new

# change s1 cls_name
s1.cls_name = "s1_cls_name"

print(zzcls.cls_name) # zzcls_cls_new
print(s1.cls_name) # s1_cls_name
print(s2.cls_name) # zzcls_cls_new

```

```python
from typing import ClassVar  # 没什么约束。最多ide帮你报个错，解释器该运行还是运行了。
# （强行）设置一个 共享的类属性  --并没有什么约束
# 但是实例属性可以覆盖类属性，实例改了这个 共享的类属性 就变成 实例属性 了。

class zzzcls:
    name: ClassVar[str] = "zzzcls"   # cls attribute
    cls_name: ClassVar[str] = "zzzcls_cls" # cls attribute  # 必须通过类名访问和修改
    
    def __init__(self, name: str):
        self.name = name   # instance attribute  

# instance attribute may cover cls attribute 
s1 = zzzcls("s1")
s2 = zzzcls("s2")
print(s1.cls_name) # zzcls_cls
print(s2.cls_name) # zzcls_cls

s1.cls_name = "s1_cls_name"
print(s1.cls_name) # s1_cls_name  # 不再共享了
print(s2.cls_name) # zzzcls_cls
print(zzzcls.cls_name) # zzzcls_cls

# 这是 ClassVar 最常见的使用场景。
# 在 dataclass 中，如果不加 ClassVar，定义在类里的变量会被自动算作 __init__ 函数的参数（即变成实例变量）。

zzzcls.cls_name = "zzzcls_cls_new"
print(zzzcls.cls_name) # zzzcls_cls_new
print(s1.cls_name) # s1_cls_name
print(s2.cls_name) # zzzcls_cls_new
```
