```python
import logging 

# 全局logging

# 1. 获取一个 Logger，通常用 __name__ 命名，这样能知道日志来自哪个文件
logger = logging.getLogger(__name__)
logger.setLevel(logging.DEBUG)  # 设置全局最低捕获级别

# 2. 创建 Handler - 输出到控制台
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)  # 控制台只看 INFO 以上

# 3. 创建 Handler - 输出到文件
file_handler = logging.FileHandler('app.log')
file_handler.setLevel(logging.DEBUG)    # 文件记录所有细节

# 4. 定义 Formatter (格式)
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# 5. 给 Handler 装上 Formatter
console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

# 6. 给 Logger 装上 Handler
logger.addHandler(console_handler)
logger.addHandler(file_handler)

# ----------------------------
# 使用
logger.info("系统启动中...")       # 输出到控制台 + 文件
logger.debug("正在连接数据库变量 x=10...") # 只输出到文件
logger.error("连接超时！")         # 输出到控制台 + 文件
```

```python
# loguru 日志配置
import sys
from loguru import logger

def setup_logger(
    log_level: str = "INFO",
    serialize: bool = False,      # True 则输出 JSON 格式，适合生产环境采集
) -> None:
    """
    整个工程调用一次，在 main.py 或服务入口处调用。
    """
    # 移除 loguru 默认的 handler，避免重复输出
    logger.remove()

    # ── 控制台输出（带颜色）────────────────────────────────
    logger.add(
        sys.stdout,
        level=log_level,
        colorize=True,
        format=(
            "<green>{time:YYYY-MM-DD HH:mm:ss}</green> | "
            "<level>{level:<8}</level> | "
            "<cyan>{name}</cyan>:<cyan>{line}</cyan> | "
            "<level>{message}</level>"
        ),
    )

setup_logger()

logger.info('info msg')
logger.warning('warning msg')
logger.error('error msg')
logger.critical('critical msg')
```
