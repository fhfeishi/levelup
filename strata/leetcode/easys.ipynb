```python
# 0001 easy twoSum

class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        m = {} 
        for i, x in enumerate(nums):
            y = target -x 
            if y in m:
                return [m[y], i]
            m[x] = i 
            
            
ss = Solution()
result = ss.twoSum([2, 7, 11, 15], 9)
print(result)  # Output: [0, 1]
```

```python
# 0002 medium addTwoNumbers


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val 
        self.next = next 
        
class Solution:
    def addTwoNumbers(
        self, 
        l1: ListNode | None, 
        l2: ListNode | None
        ) -> ListNode | None:
        
        dummy = ListNode()        
        carry, curr = 0, dummy
        while l1 or l2 or carry:
            s = (l1.val if l1 else 0) + (l2.val if l2 else 0) + carry
            carry, val = divmod(s, 10)
            curr.next = ListNode(val)
            curr = curr.next 
            
            l1 = l1.next if l1 else None 
            l2 = l2.next if l2 else None 
            
        return dummy.next     


def list2linked(nums: list[int]) -> ListNode | None:
    dummy = ListNode()
    curr = dummy 
    
    for num in nums:
        curr.next = ListNode(num)
        curr = curr.next
    return dummy.next 


def linked2list(head: ListNode | None) -> list[int]:
    nums = []
    while head:
        nums.append(head.val)
        head = head.next 
    return nums


s = Solution()
x = s.addTwoNumbers(list2linked([2,4,3]), list2linked([5,6,4]))
print(linked2list(x))





```

```python
# 0013 easy romanToInt

from itertools import pairwise

class Solution:
    def romanToInt(self, s: str) -> int:
        roman_dict = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }
        
        return sum((-1 if roman_dict[a] < roman_dict[b] else 1) * roman_dict[a] for a, b in pairwise(s)) + roman_dict[s[-1]]
        
so = Solution()
so.romanToInt("IV")
```

```python
# 0014 easy longestCommonPrefix
def longestCommonPrefix(strs: list[str]) -> str:
    if not strs: 
        return ""
    
    for i in range(len(strs[0])):
        for s in strs[1:]:
            if i == len(s) or s[i] != strs[0][i]:
                return strs[0][:i]
    
    return strs[0]

longestCommonPrefix(["flower","flow","flight"])
```

```python
# 0020 easy isValid

def isValid(s: str) -> bool:
    stack = []
    valid_pairs = {'()', '[]', '{}'}
    
    for char in s:
        if char in {'(', '[', '{'}:
            stack.append(char)
        elif char in {')', ']', '}'}:
            if not stack:
                return False
            if stack.pop() + char not in valid_pairs:
                return False
    return not stack
```
