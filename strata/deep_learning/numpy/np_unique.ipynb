````python
import numpy as np 
````

````python
arr1 = np.random.randint(0,20, size=(2,2,5))
arr1
````

````python
np.unique(arr1)
````

```python
ar：输入数组
return_index：如果为 True，返回原数组中唯一值的索引
return_inverse：如果为 True，返回一个索引数组，可用于重构原始数组
return_counts：如果为 True，返回每个唯一值在原数组中出现的次数
axis：指定沿着哪个轴寻找唯一值

````python
arr1.shape
````

````python
# axis=0    -> arr[0,:,:] arr[1,:,:] 
# axis=1    -> arr[0,0,:] arr[1,0,:]   | arr[0,1,:] arr[0,1,:]
# axis=2 -1 -> arr[01,:,?] arr[01,:,?] | arr[:,01,?] arr[:,01,?] ...
````

````python
np.unique(arr1, axis=1) 
````

````python
np.unique(arr1, axis=1).shape  
````

````python
np.unique(arr1, axis=0)       
````

````python
np.unique(arr1, axis=0).shape
````
