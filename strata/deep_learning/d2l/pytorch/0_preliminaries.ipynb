# 自动微分

````python
# Y = 2 X**2 +b
x = torch.arange(6.0)
x.requires_grad_(True)
# x.requires_grad= True 
print(f"{x.grad = }")
````

````python
y = 2 * torch.dot(x,x)
# y = 2 * torch.matmul(x,x)
y
````

````python
y.backward()
print(f"{x.grad = }")
````

````python
x.grad == 4 * x
````

````python
# 在默认情况下torch会累计梯度, 这里需要清除之前的梯度
x.grad.zero_()
y = x.sum()
y.backward()  
x.grad  
# 非标量变量的反向传播
# 反映到每个样本上  --> 每个批次内每个样本的偏导数的和
````

# python控制流的梯度计算

````python
def f(a):
    b = a*2
    while b.norm() < 1000:
        b = b*2
    if b.sum() > 0:
        c = b
    else:
        c = 100 * b
    return c 

a = torch.randn(size=(), requires_grad=True)
print(f"{a = }")
d = f(a)
d.backward()
````

````python
a.grad == d/a
````
