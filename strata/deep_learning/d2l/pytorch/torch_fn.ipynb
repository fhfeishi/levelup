````python
import torch
````

# torch.cat

```python
torch.cat(tensors, dim=0, *, out=None)

- 参数说明
    - tensors：要连接的张量序列（列表或元组）
    - dim：沿着哪个维度连接张量（默认为0）
    - out：输出张量（可选）
- 返回值
    - 返回连接后的张量。

# torch.split()

```python
torch.split(tensor, split_size_or_sections, dim=0)


- 参数说明
    - tensor：要分割的输入张量
    - split_size_or_sections：可以是整数或整数列表
        - 如果是整数，表示每个分割块的大小
        - 如果是整数列表，表示每个分割块的具体大小
    - dim：沿着哪个维度分割张量（默认为0）
- 返回值
    - 返回分割后的张量元组。

# .view()

.view() 方法用于改变张量的形状，但不改变其数据。

```python
tensor.view(shape)

- 参数说明
    - shape：新的形状，可以是一个元组或者多个整数参数
    - 可以使用 -1 表示自动计算该维度的大小
- 特点
    - 要求改变形状前后元素总数保持不变
    - 不改变张量的存储顺序（内存布局）
    - 返回的是原张量数据的不同视图（view），共享底层数据

# .permute()

.permute() 方法用于重新排列张量的维度，可以对任意高维张量进行维度重排。

- 参数说明
    - dims：新的维度顺序，需要包含所有维度的索引
- 特点
    - 可以任意重排维度顺序
    - 会创建新的底层存储，通常比 transpose() 慢但更灵活
    - 不要求维度相邻

# .transpose()

.transpose() 方法用于交换张量的两个维度。

```python
tensor.transpose(dim0, dim1)

- 参数说明
    - dim0, dim1：要交换的两个维度的索引
- 特点
    - 只交换两个指定维度，其他维度保持不变
    - 比 permute() 更高效（如果只需要交换两个维度）
    - 返回的是原张量的视图（共享底层存储）

# .softmax()
.softmax() 是将张量转换为概率分布的函数，常用于多分类模型的输出层。

```python
tensor.softmax(dim)


- 参数说明
    - dim：沿着哪个维度应用 softmax 函数
- 特点
    - 将任意实数向量转换为概率分布（所有元素非负且和为1）
    - 沿指定维度应用 softmax 函数
    - 公式：softmax(xi) = exp(xi) / Σ exp(xj)
