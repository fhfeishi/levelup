```python
# 交叉熵为什么是这样的呢？  

def loss_fn_same(outputs, targets):
    """
    outputs: [batch_size, 7, 65] 模型预测
    targets: [batch_size, 7, 65] one-hot标签
    """
    batch_size = outputs.shape[0]
    total_loss = 0
    
    # 对每个字符位置计算交叉熵损失
    for i in range(7):
        # 使用交叉熵损失(会自动处理softmax)
        loss = cross_entropy(outputs[:, i, :], targets[:, i, :].argmax(dim=1))
        total_loss += loss
    
    return total_loss / 7  # 返回平均损失
    

def loss_fn(pred, target):
    """
    pred: [B,7,65]  
    target: [B,7] (int)
    """
    loss = 0.
    for i in range(7):
        loss += cross_entropy(pred[:, i, :], target[:, i])
    return loss / 7
```
